```python
import pandas as pd
```

```python
# Clean Income column - remove $ and commas
# changing INCOME column which was originally a string to a float so that i can run a statistical summary 
df['Income'] = df['Income'].replace('[\$,]', '', regex=True).astype(float)
```

```python
#choosing the specific columns i want to use for statistics(columns with numbers)
#selecting only numerical columns
num_cols = ['Year_Birth', 'Income', 'Kidhome', 'Teenhome', 'Age',
            'NumWebPurchases', 'NumCatalogPurchases', 'NumStorePurchases',
            'NumWebVisitsMonth', 'Response_Total']
num_cols
```

```python
# See the data type of each column
print(df[num_cols].dtypes)
```

```python
# Cleaning the Income column - remove $ and commas, then fill NaN with 0 or drop
df['Income'] = df['Income'].replace('[\$,]', '', regex=True)
df['Income'] = pd.to_numeric(df['Income'], errors='coerce')
```

```python
#verifying the updated datatype of INCOME
print(df['Income'].dtype)   # should show float64
print(df['Income'].head())  # should show numbers without $
```

```python
#stats summary
#.T switches the rows and columns so that I can have the columns at the left.
stsum = df[num_cols].describe().T
stsum['Median']   = df[num_cols].median(numeric_only=True)
stsum['Skewness'] = df[num_cols].skew(numeric_only=True)

stsum
```

### Exporting this statistical summary table

```python
#excel file
stsum.to_excel('statistical_summary.xlsx', index=True)
```

```python
#csv file
stsum.to_csv('statistical_summary.csv', index=True)
```
